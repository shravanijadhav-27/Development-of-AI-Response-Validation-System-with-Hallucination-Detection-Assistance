```python
import sqlite3
import pandas as pd

# Create a temporary database
connection = sqlite3.connect(":memory:")

# Create student table
connection.execute("""
CREATE TABLE students (
    Student_ID INTEGER,
    Name TEXT,
    Department TEXT,
    Attendance INTEGER,
    Study_Hours INTEGER,
    Assignment_Marks INTEGER,
    Internal_Marks INTEGER,
    Final_Marks INTEGER
)
""")

# Add student data
students = [
    (101, "Aarav", "BCS", 85, 4, 82, 78, 80),
    (102, "Priya", "BCS", 72, 3, 70, 68, 69),
    (103, "Rahul", "BCA", 91, 6, 88, 85, 87),
    (104, "Sneha", "BCS", 68, 2, 65, 62, 64),
    (105, "Aditya", "BCA", 88, 5, 85, 82, 84),
    (106, "Neha", "BCS", 95, 7, 92, 90, 91),
    (107, "Rohan", "BCA", 76, 3, 72, 70, 71),
    (108, "Ananya", "BCS", 82, 5, 80, 76, 79),
    (109, "Vikram", "BCA", 65, 2, 60, 58, 61),
    (110, "Pooja", "BCS", 90, 6, 89, 87, 88)
]

connection.executemany(
    "INSERT INTO students VALUES (?, ?, ?, ?, ?, ?, ?, ?)",
    students
)

# Display the table
pd.read_sql_query("SELECT * FROM students", connection)
```

## Part 1: SQL Practice

Topics covered: SELECT, WHERE, ORDER BY, GROUP BY, HAVING, duplicate values, INNER JOIN, and LEFT JOIN.


```python
query = """
SELECT * FROM students;
"""
result = pd.read_sql_query(query, connection)
result
```

```python
query = """
SELECT Name, Department, Final_Marks
FROM students;
"""

result = pd.read_sql_query(query, connection)
result
```

```python
query = """
SELECT Name, Final_Marks
FROM students
ORDER BY Final_Marks DESC;
"""

result = pd.read_sql_query(query, connection)
result
```

```python
 query="""SELECT Department,AVG(Final_Marks) AS Average_Marks
From students
GROUP BY Department; """
result = pd.read_sql_query(query,connection)
result
```

```python
query = """
SELECT Department, AVG(Final_Marks) AS Average_Marks
FROM students
GROUP BY Department
HAVING AVG(Final_Marks) > 75;
"""

result = pd.read_sql_query(query, connection)
result
```

```python
connection.execute("""
CREATE TABLE departments (
    Department TEXT,
    HOD TEXT
)
""")

departments = [
    ("BCS", "Dr. Mehta"),
    ("BCA", "Dr. Sharma")
]

connection.executemany(
    "INSERT INTO departments VALUES (?, ?)",
    departments
)
```

```python
query = """
SELECT students.Name, students.Department, departments.HOD
FROM students
INNER JOIN departments
ON students.Department = departments.Department;
"""

result = pd.read_sql_query(query, connection)
result
```

```python
query = """
SELECT students.Name, students.Department, departments.HOD
FROM students
LEFT JOIN departments
ON students.Department = departments.Department;
"""

result = pd.read_sql_query(query, connection)
result
```

## Part 2: Python Practice

**Topics Covered:** Filtering data, calculating averages, finding maximum marks, counting students by department, identifying low attendance, and sorting data using Pandas.




**Prompt 1: Filter Students**

Write Python code using Pandas to display students whose final marks are greater than 80.


```python
df = pd.read_sql_query("SELECT*FROM students",connection)
result = df[df["Final_Marks"]>80]
result
```

**Prompt 2: Average Marks**

Write Python code using Pandas to calculate the average final marks of all students.


```python
average_marks=df["Final_Marks"].mean()
print("Average Final Marks:",average_marks)
```

```python
average_marks=df["Final_Marks"].mean()
print("Average Final Marks:",round (average_marks,2))
```

**Prompt 3: Highest Marks**

Write Python code using Pandas to find the student who scored the highest final marks.


```python
top_student = df.loc[df["Final_Marks"].idxmax()]
print(top_student[["Name","Final_Marks"]])
```

**Prompt 4: Department-wise Student Count**

Write Python code using Pandas to count the number of students in each department.


```python
department_count = df["Department"].value_counts()
print (department_count)
```

**Prompt 5: Low Attendance**

Write Python code using Pandas to display students whose attendance is below 75%.


```python
low_attendance = df[df["Attendance"]<75]
low_attendance
```

Write Python code using Pandas to sort students by final marks from highest to lowest.
